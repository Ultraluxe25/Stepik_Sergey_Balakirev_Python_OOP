```python
def fact(n, a=1):
    if n <= 1:
        return a

    return fact(n-1, a * n)


fact(3)
```

```python
tree_root = {'data': 1,
             'left': {'data': 2,
                      'left': {'data': 7, 'left': None, 'right': None},
                      'right': {'data': 3, 'left': None, 'right': None}
                      },
             'right': {'data': 5,
                       'left': {'data': -4, 'left': None, 'right': None},
                       'right': {'data': 8, 'left': None, 'right': None}
                       }
             }
```

```python
def find_tree_sum(node):
    if node is None:
        return 0
    
    return node['data'] + find_tree_sum(node['left']) + find_tree_sum(node['right'])
```

```python
find_tree_sum(tree_root)
```
