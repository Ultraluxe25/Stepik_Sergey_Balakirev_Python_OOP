Создалась переменная `command`, которая ссылается на cmd. В данном случае это должна быть последняя **case**, словно это `case _:`



`case str():` - шаблон проверки что переменная является строкой.



В данном случае не переменная приводится к строке, а проверяется является ли слово строкой и переменая в скобках начинает ссылаться на `cmd`



То же самое что и действие выше



Дополнительные условия можно внести после переменной и `if`.
